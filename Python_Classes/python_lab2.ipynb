- Find the mean of the entire array.

### 2. Pandas: DataFrames

```python
# Q2.1: Create a Pandas DataFrame from the following dictionary:
data = {
    'Product': ['Laptop', 'Mouse', 'Monitor', 'Keyboard'],
    'Price': [1200, 25, 300, 75],
    'Stock': [10, 50, 0, 22]

}
```

Display the first 2 rows

```python

df1 = pd.DataFrame(data)
print(df1.head(2))
```

Filter the DataFrame to show only products that are out of stock (Stock == 0).

```python
print(df1[df1["Stock"]==0])
```

- Q2.2: Using the same DataFrame:
   - Add a new column called Total_Value which is calculated by multiplying Price and Stock.

   - Calculate the total sum of the Total_Value column.

```python
df1['Total_Value'] = df1['Price'] * df1['Stock']
print(df1)
print(df1['Total_Value'].mean())
```

### 3. Data Cleaning & Analysis

```python

df = pd.DataFrame({'A': [1, 2, np.nan, 4], 'B': [5, np.nan, 7, 8]})
```

Fill the missing values (NaN) with the mean of their respective columns.

```python
df = df.fillna(df.mean())
print(df)
```

Drop the rows where the value in column 'A' is greater than 2.

```python
df.drop(df[df['A'] > 2].index, inplace=True)

print(df)
```
