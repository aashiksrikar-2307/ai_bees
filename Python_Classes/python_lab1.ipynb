### 2. Dictionaries & List Comprehension

- Q2.1: You have a list of numbers: nums = [1, 2, 3, 4, 5, 6]. Use a list comprehension to create a new list containing the square of each number, but only if the number is even.

    - Expected Output: [4, 16, 36]

```python
nums = [1,2,3,4,5,6]
nums1 = [num*num for num in nums if num%2==0 ]
print(nums1)
```

### 3. Control Flow & Loops

- Q3.1: Write a for loop that iterates from 1 to 20. If a number is divisible by 3, print "Fizz". If it's divisible by 5, print "Buzz". If it's divisible by both, print "FizzBuzz". Otherwise, print the number.

```python
for i in range(1,21):
    if i%3==0 and i%5==0:
        print("FizzBuzz")
    elif i%3==0:
        print("Fizz")
    elif i%5==0:
        print("Buzz")
    else:
        print("NA")
```

- Q3.2: Write a while loop that asks a user for a password. If the user enters "python123", print "Access Granted" and break the loop. Otherwise, print "Try again". (For Jupyter, you can simulate this by setting a variable input_str).

```python
while True:
    pas = input("Enter Password")
    if pas=="python123":
        print("Access Granted")
        break
    else:
        print("Try again")
```

### 4. Functions & Lambda

- Q4.1: Write a function calculate_area(length, width) that returns the area of a rectangle. If only one argument is provided, assume it is a square (width = length).

    - Expected Output for (5, 4): 20

    - Expected Output for (5): 25

```python
def calculate_area(length, width=None):
    if width is None:
        width = length
    return length * width

print(calculate_area(5, 4))  
print(calculate_area(5))     
```

- Q4.2: Use a lambda function and the filter() method to extract all words longer than 4 characters from this list: words = ["apple", "cat", "banana", "dog", "cherry"].

    - Expected Output: ['apple', 'banana', 'cherry']

```python
words = ["apple", "cat", "banana", "dog", "cherry"]
ab=list(filter(lambda w: len(w)>4,words))
print(ab)
```

### 6. Classes & Objects

- Q6.1: Define a class Car.

    - It should have attributes brand and speed.

    - Create a method accelerate(increment) that increases the speed.

    - Create a method status() that prints: "The [brand] is going [speed] km/h."

    - Instantiate an object of Car, accelerate it by 30, and call the status method.

```python
class Car:
    def __init__(self,brand,speed):
        self.brand = brand
        self.speed = speed
    def accelarate(self,acc):
        self.speed+=acc
    def status(self):
        print(f"The {self.brand} is going {self.speed} km/h.")
cr = Car("Audi",30)
cr.accelarate(10)
cr.status()
```
